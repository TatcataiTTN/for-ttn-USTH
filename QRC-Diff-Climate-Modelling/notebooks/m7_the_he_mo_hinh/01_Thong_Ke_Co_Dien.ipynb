# Thế hệ 1 — Thống kê/kinh nghiệm cổ điển (trước máy tính số)

Trước NWP (trước 1950), dự báo dựa trên **phân tích synoptic** (bản đồ áp suất vẽ tay, quy tắc kinh
nghiệm "nếu hình thế A thì thời tiết B") và **hồi quy thống kê tuyến tính** trên chuỗi quan trắc lịch
sử — không giải phương trình vật lý, chỉ khớp quan hệ thống kê.

Demo dưới đây dùng đúng dữ liệu SST thật (NOAA OISST v2.1, vùng Biển Đông 5-25°N, 100-120°E, 12
tháng năm 2020) để minh hoạ: dự báo tháng sau chỉ bằng **hồi quy tuyến tính trên chính nó (persistence
+ linear regression)** — đúng kiểu mô hình "thế hệ 1", không dùng phương trình động lực nào cả.

```python
DATA_PATH = r"/Users/tuannghiat/Downloads/USTH Phenikaa - Climate Modelling /08_The_He_Mo_Hinh_Thoi_Tiet/_data_cache/oisst_SEA_2020_monthly_sample.nc"
print("Dùng dữ liệu thật:", DATA_PATH)
```

```python
import xarray as xr
import numpy as np

ds = xr.open_dataset(DATA_PATH)
sst_series = ds['sst'].mean(dim=['lat', 'lon']).squeeze().values
months = ds['time'].values
print("SST trung bình vùng Biển Đông theo tháng (độ C), dữ liệu thật OISST v2.1 2020:")
for m, v in zip(months, sst_series):
    print(str(m)[:7], round(float(v), 3))
```

```python
# Hồi quy tuyến tính cổ điển: SST(t+1) ~ a*SST(t) + b  (không có vật lý, chỉ thống kê)
x = sst_series[:-1]
y = sst_series[1:]
a, b = np.polyfit(x, y, 1)
pred = a * x + b
resid = y - pred
rmse = np.sqrt(np.mean(resid**2))
print(f"Hồi quy thống kê cổ điển: SST(t+1) = {a:.3f} * SST(t) + {b:.3f}")
print(f"RMSE trên chính dữ liệu fit (in-sample, chỉ minh hoạ): {rmse:.3f} độ C")
print("Hạn chế thế hệ này: không có phương trình động lực/vật lý nào, chỉ ngoại suy thống kê —")
print("không thể dự báo khi hình thế thời tiết chưa từng xảy ra trong lịch sử quan trắc.")
```
