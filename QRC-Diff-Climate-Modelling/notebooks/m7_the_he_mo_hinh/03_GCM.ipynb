# Thế hệ 3 — General Circulation Models (GCM)

GCM mở rộng mô hình NWP nguyên thuỷ sang **toàn cầu, nhiều tầng khí quyển**, chạy dài hạn (mùa/năm
thay vì vài ngày), và phải dùng **tham số hoá vật lý** (parameterization) cho các quá trình nhỏ hơn
lưới tính (mây, đối lưu, bức xạ) vì độ phân giải lưới quá thô (thường >100km) để giải trực tiếp.

Demo dùng dữ liệu SST thật (OISST v2.1 vùng Biển Đông, 2020) để minh hoạ đúng vấn đề **độ phân giải
thô** của GCM: lấy lưới 0.25° thật rồi coarsen (lấy trung bình khối) xuống độ phân giải kiểu GCM
(~2°, gần giống lưới CMIP6 điển hình) — cho thấy trực quan vì sao cần "downscaling" ở các thế hệ sau.

```python
DATA_PATH = r"/Users/tuannghiat/Downloads/USTH Phenikaa - Climate Modelling /08_The_He_Mo_Hinh_Thoi_Tiet/_data_cache/oisst_SEA_2020_monthly_sample.nc"
print("Dùng dữ liệu thật:", DATA_PATH)
```

```python
import xarray as xr
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

ds = xr.open_dataset(DATA_PATH)
sst_jul = ds['sst'].isel(time=6).squeeze()  # tháng 7/2020, cao điểm gió mùa Tây Nam
print("Lưới gốc (độ phân giải OISST 0.25°):", sst_jul.shape)

# Coarsen về độ phân giải kiểu GCM (~2 độ = 8 ô lưới 0.25 độ gộp lại)
sst_coarse = sst_jul.coarsen(lat=8, lon=8, boundary='trim').mean()
print("Lưới sau coarsen kiểu GCM (~2 độ):", sst_coarse.shape)

fig, axes = plt.subplots(1, 2, figsize=(10,4))
sst_jul.plot(ax=axes[0], cmap='turbo'); axes[0].set_title('SST thật, 0.25° (như vệ tinh quan trắc)')
sst_coarse.plot(ax=axes[1], cmap='turbo'); axes[1].set_title('SST coarsen ~2° (độ phân giải kiểu GCM)')
plt.tight_layout(); plt.savefig('gcm_coarsen_comparison.png', dpi=100)
print("Chênh lệch SST lớn nhất giữa 2 lưới tại cùng vị trí xấp xỉ (do làm mượt chi tiết nhỏ):",
      round(float(np.nanmax(np.abs(sst_jul.values - sst_coarse.interp_like(sst_jul, method='nearest').values))), 3), "độ C")
print("Đây chính là lý do thế hệ GCM không thể hiện được chi tiết vùng ven biển/đảo nhỏ —")
print("động lực ra đời của RCM/downscaling (thế hệ 5-6) để khôi phục chi tiết không gian.")
```
